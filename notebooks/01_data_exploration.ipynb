```python
def profile_dataframe(data):
    print("Dimensao: \n", data.shape)
    print("Colunas: \n",data.columns)
    print("Valores NUllos \n: ", data.null_count)
    print("Linhas unicas: \n", data.n_unique)
    print("Valores distintos \n", data.select(polas.all().n_unique()))
```

```python
profile_dataframe(bridge_job_skills)
```

```python
profile_dataframe(dim_company)
```

```python
fact_job_postings.select(
    polas.col("job_title").value_counts(sort=True)
)
```

```python
job_title_counts = (
    fact_job_postings
    .select(polas.col("job_title").value_counts(sort=True))
    .unnest("job_title")
)

job_title_counts
```

```python
company_counts = (
    fact_job_postings
    .select(
        polas.col("company").value_counts(
            sort=True,
            name="count"
        )
    )
    .unnest("company")
)

company_counts
```

```python
jobs_by_country_and_level = (
    fact_job_postings
    .group_by(["country", "experience_level"])
    .agg(polas.len().alias("job_count"))
    .sort("job_count", descending=True))

jobs_by_country_and_level
```
